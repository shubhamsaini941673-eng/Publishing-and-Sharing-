Q1. What is the key difference between publishing a dashboard on Tableau Public versus Tableau Server/
Cloud in terms of accessibility and data security?

```python
The key difference is that dashboards published on Tableau Public are openly accessible to anyone on the internet with no data security controls, while dashboards on Tableau Server/Cloud are restricted to authorized users within an organization and protected by enterprise-grade security and compliance standards.

1. Accessibility

 1. Tableau Public

  . Dashboards are publicly visible worldwide.

  . Anyone with the link can view and download the workbook.

  . No authentication or user-level permissions.

  . Best suited for sharing non-sensitive, educational, or portfolio content.

 2. Tableau Server/Cloud

  . Dashboards are restricted to licensed users within the organization.

  . Supports role-based access control and integration with identity providers (LDAP, SAML, etc.).

  . Enables collaboration in a secure environment.

  . Ideal for internal analytics, business intelligence, and regulated industries.

2. Data Security

 1. Tableau Public

  . No encryption or compliance guarantees.

  . Data is uploaded to Tableau’s public servers, outside organizational control.

  . Risk: Sensitive or proprietary data becomes exposed.

 2. Tableau Server

  . Self-hosted, giving organizations full control over infrastructure, security protocols, and compliance.

  . Customizable to meet strict regulatory requirements (HIPAA, GDPR, etc.).

  . Suitable for organizations needing data residency control.

 3. Tableau Cloud

  . Fully managed SaaS platform by Tableau.

  . Provides enterprise-grade security, ISO/SOC certifications, GDPR compliance, and strong authentication.

  . Tableau employees cannot access customer data unless explicitly permitted.

  . Lower IT overhead compared to Server, but less direct infrastructure control.

3. Comparison Tables

| **Aspect** | **[Tableau Public](ca://s?q=Tableau_Public_vs_Server)** | **[Tableau Server](ca://s?q=Tableau_Server_features)** | **[Tableau Cloud](ca://s?q=Tableau_Cloud_features)** |
| --- | --- | --- | --- |
| **Accessibility** | Open to all | Restricted to org users | Restricted to org users |
| **Security** | None | Full organizational control | Managed by Tableau |
| **Compliance** | Not supported | Customizable | ISO, SOC, GDPR |
| **Best Use Case** | Public sharing, portfolios | Regulated industries, strict control | Agile teams, lower IT overhead |

4. Risks & Considerations

 . Publishing sensitive dashboards on Tableau Public can lead to data leaks and compliance violations.

 . Tableau Server requires significant IT resources for maintenance and upgrades.

 . Tableau Cloud reduces IT burden but may not meet strict residency or custom security needs.

. Bottom Line: Use Tableau Public only for non-sensitive, educational, or promotional dashboards. For business-critical analytics, choose Tableau Server if you need maximum control and compliance, or Tableau Cloud if you prefer a secure, managed SaaS solution with lower overhead.
```

Q2. When publishing a Tableau dashboard, what happens to a live connection compared to an extract? How does this affect data refresh and performance?

```python
. When you publish a Tableau dashboard, a live connection keeps querying the underlying database in real time, while an extract stores a compressed snapshot of the data in Tableau’s Hyper engine. This means live connections provide up-to-the-minute freshness but often suffer from slower performance, while extracts deliver faster, more responsive dashboards but require scheduled refreshes to stay current.

1. Live Connection vs Extract

 1. Live Connection

  . Behavior: Every filter, sort, or interaction triggers a query directly against the database.

  . Freshness: Always reflects the latest data available in the source.

  . Performance Impact: Can be slow if the database is large, under heavy load, or not optimized for interactive queries.

  . Best Use Cases: Real-time monitoring dashboards (fraud detection, trading, incident response).

. Extract

  . Behavior: Data is stored locally in Tableau’s Hyper engine as a compressed snapshot.

  . Freshness: Depends on the last scheduled refresh (hourly, daily, weekly).

  . Performance Impact: Much faster because queries run in-memory, optimized for dashboard interactivity.

  . Best Use Cases: Business dashboards viewed periodically (sales reports, executive summaries).

2. Comparison Tables

| **Aspect** | **[Live Connection](ca://s?q=Tableau_live_connection_vs_extract)** | **[Extract](ca://s?q=Tableau_extract_vs_live_connection)** |
| --- | --- | --- |
| **Data Freshness** | Real-time, always current | Snapshot, depends on refresh |
| **Performance** | Slower, database-dependent | Faster, in-memory optimized |
| **Database Load** | High (every interaction = query) | Low (queries run locally) |
| **Best Use Case** | Real-time dashboards | Scheduled reporting & analytics |

3. Key Considerations

 . Performance Trade-off: Extracts consistently outperform live connections by ~12 points in Tableau’s performance health scores, making dashboards feel “snappy” instead of sluggish.

 . Refresh Strategy: Extracts require scheduled refreshes to balance freshness with speed. Hourly or nightly refreshes often suffice for most business dashboards.

 . Database Strain: Live connections can overwhelm production databases if many users interact simultaneously. Extracts reduce this risk by offloading queries to Tableau’s Hyper engine.

 4. Buttom Line

 . Use live connections only when dashboards truly need real-time data.

 . For most business scenarios, extracts are the better choice — they deliver faster performance, reduce database strain, and can be refreshed on a schedule that matches business needs.
```

Q3. In Tableau Server or Cloud (conceptually), which feature allows administrators to control which users can view, interact with, or edit a published dashboard?

```python
. In Tableau Server or Tableau Cloud, administrators control user access through permissions and site roles. These features define who can view, interact with, or edit a published dashboard.

1. Key Feartures

 1. Permissoion

. Granular control at the workbook, dashboard, or data source level.

. Define actions like:

  . View (read-only access)

  . Interact (apply filters, drill down, export data)

  . Edit/Save (modify dashboards or publish changes)

. Permissions can be customized per user or group.

 2. Site Roles

. Broad access levels assigned to users across the site.

Examples:

  . Viewer → Can only see dashboards.

  . Explorer → Can interact and create new content from published data.

  . Creator → Full authoring rights, including publishing new dashboards.

. Site roles set the baseline capabilities, while permissions refine them.

2. How They Work Together

| **Feature** | **Purpose** | **Example Use Case** |
| --- | --- | --- |
| **[Site Roles](ca://s?q=Tableau_site_roles_vs_permissions)** | Define overall capabilities | A business analyst is an *Explorer* |
| **[Permissions](ca://s?q=Tableau_permissions_examples)** | Fine-tune access per dashboard/workbook | Analyst can view Sales Dashboard but cannot edit HR Dashboard |

3. Buttom Line

Administrators use site roles to establish a user’s general level of access, and then apply permissions to control what each user or group can do with specific dashboards. Together, they ensure secure, role-based access to published content.
```

Q4. If you want users to view a dashboard but not download the underlying data, which permission setting would you adjust in Tableau Server?

```python
. To prevent users from downloading the underlying data while still allowing them to view a dashboard in Tableau Server or Tableau Cloud, you would adjust the Download Data permission setting.

1. How It Works

. Download Data → Controls whether users can export the underlying data behind a visualization.

. If unchecked/denied, users can still view and interact with the dashboard but cannot:

  . Export data to CSV/Excel.

  . Access raw tables behind charts.

. Other related permissions include:

  . Download Workbook/Save a Copy → Prevents users from downloading the entire workbook.

  . Export Image/PDF → Allows exporting the visualization itself, but not the raw data.

2. Example Scenario

| **Permission** | **Allowed** | **Denied** |
| --- | --- | --- |
| **View** | User can open and see dashboards | — |
| **Interact** | User can filter, drill down | — |
| **Download Data** | User can export raw data | ❌ Prevented |
| **Export Image/PDF** | User can save a snapshot | ✅ Allowed |

. Bottom Line: To let users view dashboards but block access to raw data, deny the "Download Data" permission while keeping View and Interact enabled.
```

Q5. Name two methods by which a Tableau dashboard can be embedded into a web page or website?

```python
. Two common methods to embed a Tableau dashboard into a web page or website are:

1. Embedding Methods

 1. Embed Code (iFrame)

  . Tableau provides an Embed Code option when publishing a dashboard.

  . This generates an HTML <iframe> snippet that can be pasted directly into a web page.

  . Simple and widely supported, but limited in customization.

 2. JavaScript API

  . Tableau’s JavaScript API allows developers to embed dashboards with interactive controls.

  . Enables advanced features like:

   . Passing parameters from the website to Tableau.

   . Controlling filters and views dynamically.

   . Listening for user interactions (e.g., filter changes).

  . More flexible than iFrame embedding, ideal for integrated web applications.

2. Comparison

| **Method** | **Ease of Use** | **Customization** | **Best Use Case** |
| --- | --- | --- | --- |
| **[iFrame Embed](ca://s?q=Tableau_iframe_vs_js_api)** | Very easy | Limited | Quick embedding in blogs, portals |
| **[JavaScript API](ca://s?q=Tableau_js_api_vs_iframe)** | Requires coding | Highly customizable | Enterprise apps, interactive sites |

Bottom Line: Use iFrame embedding for quick, simple publishing, and the JavaScript API when you need deeper integration and control over dashboard behavior.
```

Q6. List two best practices you should follow when preparing a dashboard for public sharing to ensure data security and privacy.

```python
. Two best practices to follow when preparing a Tableau dashboard for public sharing to ensure data security and privacy are:

1. Best Practices

 1. Remove sensitive data

  . Strip out confidential fields such as customer names, financial details, or personal identifiers.

  . Replace with aggregated or anonymized values (e.g., show totals or percentages instead of raw records).

  . This prevents accidental exposure of proprietary or personal information.

 2. Apply data anonymization and aggregation

  . Use calculated fields or summary tables to present insights without revealing row-level data.

  . Aggregate at higher levels (e.g., region instead of individual city, age group instead of exact age).

  . Ensures privacy while still delivering meaningful analytics.

2. Scenario

. Imagine publishing a sales dashboard:

  . Instead of showing customer-level transactions, only display regional sales totals.

  . Instead of exposing employee IDs, show department-level performance metrics.

.Bottom Line: Before sharing dashboards publicly, always remove sensitive fields and aggregate/anonymize data to protect privacy while maintaining analytical value.
```

Q7.  report a “Data not found” error.
 What is the most common cause related to data source settings?

```python
. The most common cause of a “Data not found” error in Tableau, when related to data source settings, is a broken or incorrect connection path.

1. Typical Causes

 1. Invalid file path or database connection

  . The workbook points to a local file or database that has been moved, renamed, or is no longer accessible.

  . Example: A CSV file was stored on a desktop but isn’t available on Tableau Server.

 2. Authentication or credential issues

  . Saved credentials may have expired or not been embedded when publishing.

  . Example: A SQL Server connection requires a password, but the workbook was published without embedding credentials.

2. Affects Dashboards

| **Cause** | **Impact** |
| --- | --- |
| **Broken path** | Tableau cannot locate the data file or database → “Data not found” |
| **Missing credentials** | Tableau cannot authenticate → “Data not found” |
| **Permission mismatch** | User lacks access rights → Dashboard loads but data is missing |

 Bottom Line: The most common culprit is that the data source connection settings (path or credentials) are invalid or unavailable after publishing. Ensuring embedded credentials or using server-accessible data sources usually resolves this.
```

Q8. When embedding a Tableau dashboard into a secure internal portal, what two additional security measures should be implemented to ensure only authorized users can access the embedded content?

```python
. When embedding a Tableau dashboard into a secure internal portal, two additional security measures should be implemented to ensure only authorized users can access the content:

1. Security Measures

 . Single Sign-On (SSO) Integration

 . Connect Tableau Server/Cloud with your organization’s identity provider (e.g., Active Directory, SAML, OAuth).

 . Ensures that only authenticated employees can access the embedded dashboard.

 . Provides seamless login experience without requiring separate Tableau credentials.

2. Row-Level Security (RLS)

 . Restricts the data each user can see based on their role, department, or access rights.

 . Example: A sales manager sees only their region’s data, while executives see company-wide data.

 . Prevents unauthorized exposure of sensitive information even within the same dashboard.

3. Combined Effect

| **Measure** | **Purpose** | **Benefit** |
| --- | --- | --- |
| **[SSO Integration](ca://s?q=Tableau_SSO_security_benefits)** | Authenticate users via corporate identity provider | Ensures only employees can access |
| **[Row-Level Security](ca://s?q=Tableau_RLS_security_benefits)** | Filter data dynamically per user/group | Protects sensitive subsets of data |

Bottom Line: Use SSO to control who can access the embedded dashboard, and RLS to control what data each authorized user can see. Together, they provide strong, layered security for internal portals.
```
